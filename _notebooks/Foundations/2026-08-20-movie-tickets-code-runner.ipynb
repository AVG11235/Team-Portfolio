## Pseudocode Code Cells

Reference for the College Board AP CSP pseudocode used below: `←` for assignment, `DISPLAY(...)` to print, `INPUT(...)` to read a value, `IF/ELSE`, `REPEAT n TIMES`, `REPEAT UNTIL`, `FOR EACH item IN list`, `PROCEDURE name(params)`, `RETURN`, and `MOD` for remainder.

## Practical Example of Output: Movie Ticket Receipt

```python
%%pseudocode

// CODE_RUNNER: Shows receipt for movie ticket

// ============================
// OUTPUT: Display ticket order details
// ============================

movieTitle ← "Ponyo"
ticketPrice ← 12.50
ticketCount ← 3

DISPLAY("=== Movie Ticket Receipt ===")
DISPLAY("Movie: " + movieTitle)
DISPLAY("Price per ticket: $" + ticketPrice)
DISPLAY("Number of tickets: " + ticketCount)
DISPLAY("Thank you for your purchase!")
```

## Practical Example of INPUT: Movie Ticket Customer Discount

```python
%%pseudocode

// CODE_RUNNER: Personalize your movie ticket order.

// ============================
// INPUT: Collect customer data
// ============================

DISPLAY("=== Movie Ticket Discount Calculator ===")

customerName ← INPUT("Enter your name:")
movieType ← INPUT("What movie are you viewing?")
customerAge ← INPUT("Enter your age:")
ticketPrice ← 12.50 // fixed value for simplicity

// ============================
// Determine discount based on age
// ============================

discount ← 0

IF (customerAge < 13)
{
  discount ← 0.30        // child discount
}
ELSE
{
  IF (customerAge >= 65)
  {
    discount ← 0.25      // senior discount
  }
  ELSE
  {
    IF (customerAge >= 13 AND customerAge <= 17)
    {
      discount ← 0.15    // teen discount
    }
    ELSE
    {
      discount ← 0        // full price, adult
    }
  }
}

finalPrice ← ticketPrice - (ticketPrice * discount)

DISPLAY("Hello, " + customerName + "!")
DISPLAY("Age: " + customerAge)
DISPLAY("Movie: " + movieType)
DISPLAY("Discount applied: " + (discount * 100) + "%")
DISPLAY("Your ticket price: $" + finalPrice)
DISPLAY("Original Price: $" + ticketPrice)
```

## Practical Example of LIST: Movie Ticket List and Prices

```python
%%pseudocode

// CODE_RUNNER: Store multiple movies and prices in a list and display them.

// ============================
// LIST: Store multiple movies and prices
// ============================

movieTitles ← ["Dune: Part Three", "Ponyo", "The Wild Robot", "Wall-E"]
moviePrices ← [12.50, 14.00, 11.75, 13.25]

DISPLAY("=== Now Showing ===")

FOR EACH title IN movieTitles
{
  DISPLAY(title)
}

// Accessing a specific item by index (index starts at 1)
firstMovie ← movieTitles[1]
firstPrice ← moviePrices[1]

DISPLAY("Featured movie: " + firstMovie)
DISPLAY("Featured price: $" + firstPrice)

DISPLAY("Total movies playing: " + LENGTH(movieTitles))
```

## Practical Example of PROCEDURE: Ticket Total Price Procedure

```python
%%pseudocode

// CODE_RUNNER: Complete the procedure so it returns the discounted price. A student discount takes 20% off.

// ============================
// PROCEDURE: Calculate ticket total
// ============================

PROCEDURE calculateTotal(price, ticketCount)
{
  total ← price * ticketCount
  RETURN(total)
}

// ---- Calling the procedure ----

ticketPrice ← 12.50
numberOfTickets ← 3

orderTotal ← calculateTotal(ticketPrice, numberOfTickets)

DISPLAY("Price per ticket: $" + ticketPrice)
DISPLAY("Tickets purchased: " + numberOfTickets)
DISPLAY("Order total: $" + orderTotal)
```

## Practical Example of Sequence: Calculate Total Cost

## Practical Example of Selection: Apply Matinee or Senior Discount

```python
%% pseudocode

// CODE_RUNNER: Calculate the discounted price for a movie ticket based on age and matinee status.

age <- 68
isMatinee <- true
price <- 15

IF (age >= 65 OR isMatinee)
{
    price <- price - 5
    DISPLAY ("Discount applied!")
}
ELSE
{
    DISPLAY ("Regular price charged.")
}

DISPLAY (price)
```

## Practical Example of Iteration: Assign Multiple Seats

```python
%% pseudocode

// CODE_RUNNER: Assign seats for a movie theater based on the number of tickets purchased.

ticketsNeeded <- 3
seatsAssigned <- 0
seatNumber <- 1

REPEAT UNTIL (seatsAssigned = ticketsNeeded)
{
    DISPLAY ("Seat ")
    DISPLAY (seatNumber)
    DISPLAY (" assigned.")
    
    seatNumber <- seatNumber + 1
    seatsAssigned <- seatsAssigned + 1
}
```

## Practical Example of Algorithm: Determine Final Ticket Price

```python
%% pseudocode

// CODE_RUNNER: Complete the procedure to calculate the ticket price based on age, time of day, and whether the movie is in 3D.

PROCEDURE getTicketPrice (age, time, is3D)
{
    basePrice <- 15
    
    IF (time < 16)
    {
        basePrice <- basePrice - 3
    }
    
    IF (age <= 12 OR age >= 65)
    {
        basePrice <- basePrice - 4
    }
    
    IF (is3D)
    {
        basePrice <- basePrice + 5
    }
    
    RETURN basePrice
}
```

## Practical Example of List Operations: Manage Movie Roster

```python
%% pseudocode

// CODE_RUNNER: Store multiple movies in a list and perform operations on the list.

nowPlaying <- ["Inception", "The Matrix", "Interstellar"]

APPEND (nowPlaying, "Dune")        // Adds "Dune" to the end
REMOVE (nowPlaying, 2)             // Removes "The Matrix"
INSERT (nowPlaying, 1, "Avatar")   // Inserts "Avatar" at the beginning

totalMovies <- LENGTH (nowPlaying)
DISPLAY (totalMovies)
```

## Practical Example of Search Algorithm: Check if a Movie is Playing

```python
%% pseudocode

// CODE_RUNNER: Check if a specific movie is currently playing and display an appropriate message.

nowPlaying <- ["Avatar", "Inception", "Interstellar", "Dune"]
targetMovie <- "Dune"
isFound <- false

FOR EACH movie IN nowPlaying
{
    IF (movie = targetMovie)
    {
        isFound <- true
    }
}

IF (isFound)
{
    DISPLAY ("Tickets are available for this movie.")
}
ELSE
{
    DISPLAY ("Sorry, this movie is not currently playing.")
}
```

## Practical Example of Boolean Logic: Check Promo Code Eligibility

```python
%% pseudocode

// CODE_RUNNER: Determine if a user is eligible for a discount based on their status and the day of the week.

isStudent <- true
isTuesday <- false
hasVipPass <- true

// A user gets a discount if it's Tuesday and they are a student, OR if they have a VIP pass
IF ((isStudent AND isTuesday) OR hasVipPass)
{
    DISPLAY ("Promo code accepted!")
}
ELSE
{
    DISPLAY ("Not eligible for promo.")
}
```
