## Pseudocode Code Cells

Reference for the College Board AP CSP pseudocode used below: `←` for assignment, `DISPLAY(...)` to print, `INPUT(...)` to read a value, `IF/ELSE`, `REPEAT n TIMES`, `REPEAT UNTIL`, `FOR EACH item IN list`, `PROCEDURE name(params)`, `RETURN`, and `MOD` for remainder.

## Practical Example of Output: Movie Ticket Receipt

A simple movie ticket receipt system using output with CB pseudocode variables.

```python
%%pseudocode

// CODE_RUNNER: Shows receipt for movie ticket

// ============================
// OUTPUT: Display ticket order details
// ============================

movieTitle ← "Ponyo"
ticketPrice ← 12.50
ticketCount ← 3

DISPLAY("=== Movie Ticket Receipt ===")
DISPLAY("Movie: " + movieTitle)
DISPLAY("Price per ticket: $" + ticketPrice)
DISPLAY("Number of tickets: " + ticketCount)
DISPLAY("Thank you for your purchase!")
```

## Practical Example of INPUT: Movie Ticket Customer Discount

A simple movie ticket discount system using inputs with CB pseudocode variables.

```python
%%pseudocode

// CODE_RUNNER: Personalize your movie ticket order.

// ============================
// INPUT: Collect customer data
// ============================

DISPLAY("=== Movie Ticket Discount Calculator ===")

customerName ← INPUT("Enter your name:")
movieType ← INPUT("What movie are you viewing?")
customerAge ← INPUT("Enter your age:")
ticketPrice ← 12.50 // fixed value for simplicity

// ============================
// Determine discount based on age
// ============================

discount ← 0

IF (customerAge < 13)
{
  discount ← 0.30        // child discount
}
ELSE
{
  IF (customerAge >= 65)
  {
    discount ← 0.25      // senior discount
  }
  ELSE
  {
    IF (customerAge >= 13 AND customerAge <= 17)
    {
      discount ← 0.15    // teen discount
    }
    ELSE
    {
      discount ← 0        // full price, adult
    }
  }
}

finalPrice ← ticketPrice - (ticketPrice * discount)

DISPLAY("Hello, " + customerName + "!")
DISPLAY("Age: " + customerAge)
DISPLAY("Movie: " + movieType)
DISPLAY("Discount applied: " + (discount * 100) + "%")
DISPLAY("Your ticket price: $" + finalPrice)
DISPLAY("Original Price: $" + ticketPrice)
```

## Practical Example of LIST: Movie Ticket List and Prices

A pseudocode example of a list of movies and their prices.

```python
%%pseudocode

// ============================
// LIST: Store multiple movies and prices
// ============================

movieTitles ← ["Dune: Part Three", "Ponyo", "The Wild Robot", "Wall-E"]
moviePrices ← [12.50, 14.00, 11.75, 13.25]

DISPLAY("=== Now Showing ===")

FOR EACH title IN movieTitles
{
  DISPLAY(title)
}

// Accessing a specific item by index (index starts at 1)
firstMovie ← movieTitles[1]
firstPrice ← moviePrices[1]

DISPLAY("Featured movie: " + firstMovie)
DISPLAY("Featured price: $" + firstPrice)

DISPLAY("Total movies playing: " + LENGTH(movieTitles))
```

## Practical Example of PROCEDURE: Ticket Total Price Procedure

A pseudocode example that finds the total price from the amount of tickets and price of them

```python
%%pseudocode

// CODE_RUNNER: Complete the procedure so it returns the discounted price. A student discount takes 20% off.

// ============================
// PROCEDURE: Calculate ticket total
// ============================

PROCEDURE calculateTotal(price, ticketCount)
{
  total ← price * ticketCount
  RETURN(total)
}

// ---- Calling the procedure ----

ticketPrice ← 12.50
numberOfTickets ← 3

orderTotal ← calculateTotal(ticketPrice, numberOfTickets)

DISPLAY("Price per ticket: $" + ticketPrice)
DISPLAY("Tickets purchased: " + numberOfTickets)
DISPLAY("Order total: $" + orderTotal)
```
